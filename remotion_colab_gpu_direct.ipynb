### Step 4: Verify GPU Acceleration in Remotion
Tests that Chromium launches with NVIDIA ANGLE/EGL hardware acceleration without root sandbox errors.

```python
%cd /content/repo/my-video
!npx remotion gpu \
    --gl=angle-egl \
    --chromium-options="--no-sandbox,--disable-setuid-sandbox,--disable-dev-shm-usage,--use-gl=angle,--use-angle=gl-egl"
```

### Step 5: Render Full Video with GPU Acceleration

```python
%cd /content/repo/my-video
# Direct full GPU render (8325 frames)
!npx remotion render src/index.ts AkoAngBagyo /content/ako-ang-bagyo-gpu.mp4 \
    --gl=angle-egl \
    --chromium-options="--no-sandbox,--disable-setuid-sandbox,--disable-dev-shm-usage,--use-gl=angle,--use-angle=gl-egl" \
    --concurrency=100% \
    --timeout=180000
```
